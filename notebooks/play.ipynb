```python
import numpy as np

weeks_dict = {}

for n_weeks in range(1, 52):  # From 1 to 51 representative weeks
    weeks = np.linspace(2, 52, n_weeks, dtype=int)
    weeks = sorted(set(weeks))  # Ensure uniqueness if any duplicates due to rounding
    weeks_dict[len(weeks)] = weeks  # Key is actual number of unique weeks

# Optional: print nicely
for k, v in weeks_dict.items():
    print(f"{k}: {v}")
```

```python
def generate_evenly_spaced_weeks(n_weeks):
    return sorted(
        set(int(round(w)) for w in np.linspace(1, 52, n_weeks + 2)[1:])  # exclude edges
    )


# Reasonable subset of week counts to test with
selected_week_counts = [2, 4, 6, 8, 12, 16, 26, 36, 52]

weeks_subset = {n: generate_evenly_spaced_weeks(n) for n in selected_week_counts}

# Optional: Print the dictionary
for n, weeks in weeks_subset.items():
    print(f"{n}: {weeks}")
```
